When searching for an API related to Pokemon I actually came across a Reddit thread where someone, out of free will, created 8 different surveys to figure out what the least chosen Pokemon were when playing the games. The surverys were divided up by the different regions in the world (Kanto, Johto, Hoenn, Sinnoh, Unova, NewUnova, Kalos, and Alola) and the respondants were asked to select all the Pokemon that they used at least once in any main stream game. So there wasn't an absurd amount of Pokemon listed, the creator of these surveys listed all of them as their final evolution. They clarified to the respondants that if they never reached a Pokemon's final evolution, but still used it, to select it. Overall they had about 7,000 responses for the two days the surveys were up in 2017.

The data accesible through these surveys is the generation, number, and the type(s) of each Pokemon. The numbers below the region represnt how many people who took that singular survey selected that Pokemon. Then the column to the right has the total number of respondants as the header and the percentange conversion below. The last three columns of the chart is the total number, respondants, and percentages of all the surverys put together.
